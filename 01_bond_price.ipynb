```python
def bond_price(face, coupon_rate, ytm, years, freq=2):
    """
    Price of a plain fixed-rate bond.
    face: face value (e.g. 1000)
    coupon_rate: annual coupon as decimal (0.08 = 8%)
    ytm: required annual yield as decimal
    years: years to maturity
    freq: coupon payments per year (2 = semiannual)
    """
    n = int(round(years * freq))      # total number of periods
    c = face * coupon_rate / freq     # coupon per period
    y = ytm / freq                    # yield per period

    pv_coupons = sum(c / (1 + y) ** t for t in range(1, n + 1))
    pv_face = face / (1 + y) ** n
    return pv_coupons + pv_face
```

```python
print(bond_price(1000, 0.08, 0.10, 5))   # expect ≈ 922.78
print(bond_price(1000, 0.08, 0.08, 5))   # expect 1000
print(bond_price(1000, 0.08, 0.00, 5))   # expect 1400

assert abs(bond_price(1000, 0.08, 0.10, 5) - 922.78) < 0.01
assert abs(bond_price(1000, 0.08, 0.08, 5) - 1000) < 1e-6
assert abs(bond_price(1000, 0.08, 0.00, 5) - 1400) < 1e-6
print("All tests passed")
```
