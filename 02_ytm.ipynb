```python
def bond_price(face, coupon_rate, ytm, years, freq=2):
    n = int(round(years * freq))
    c = face * coupon_rate / freq
    y = ytm / freq
    pv_coupons = sum(c / (1 + y) ** t for t in range(1, n + 1))
    pv_face = face / (1 + y) ** n
    return pv_coupons + pv_face


def find_ytm(price, face, coupon_rate, years, freq=2, tol=0.0001, max_iter=200):
    low, high = 0.0, 1.0
    for _ in range(max_iter):
        mid = (low + high) / 2
        p = bond_price(face, coupon_rate, mid, years, freq)
        if abs(p - price) < tol:
            return mid
        if p > price:
            low = mid
        else:
            high = mid
    return mid
```

```python
print(find_ytm(922.7826507, 1000, 0.08, 5))
print(find_ytm(1000, 1000, 0.08, 5))
print(find_ytm(1081.11, 1000, 0.08, 5))
```

```python
for y in [0.03, 0.05, 0.07, 0.09, 0.12]:
    p = bond_price(1000, 0.08, y, 5)
    result = find_ytm(p, 1000, 0.08, 5)
    print(y, result)
    assert abs(result - y) < 1e-5
print("Round-trip passed")
```
