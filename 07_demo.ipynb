```python
import ipywidgets as widgets
from IPython.display import display
from bond_toolkit import *

face = widgets.FloatText(value=1000, description="Face")
coupon = widgets.FloatSlider(value=8, min=0, max=15, step=0.25, description="Coupon %")
years = widgets.FloatSlider(value=5, min=1, max=30, step=1, description="Years")
yld = widgets.FloatSlider(value=10, min=0.5, max=20, step=0.25, description="Yield %")

def show(f, c, n, y):
    p = bond_price(f, c / 100, y / 100, n)
    mac, mod = durations(f, c / 100, y / 100, n)
    cv = convexity(f, c / 100, y / 100, n)
    print(f"Price: {p:.2f}")
    print(f"Macaulay duration: {mac:.3f} years")
    print(f"Modified duration: {mod:.3f}")
    print(f"Convexity: {cv:.2f}")

out = widgets.interactive_output(show, {"f": face, "c": coupon, "n": years, "y": yld})
display(face, coupon, years, yld, out)
```
